```python
# ==========================================
# Quantitative Finance Lab
# Day 1 - Financial Mathematics
# ==========================================


# Future Value
def future_value(pv, r, n):
    """
    Calculate future value.

    pv = present value
    r = annual interest rate
    n = number of years
    """
    return pv * (1 + r) ** n


# Present Value
def present_value(fv, r, n):
    """
    Calculate present value.

    fv = future value
    r = discount rate
    n = number of years
    """
    return fv / (1 + r) ** n


# Net Present Value
def npv(rate, cash_flows):
    """
    Calculate net present value.

    rate = discount rate
    cash_flows = list of cash flows
    """
    total = 0

    for year, cash_flow in enumerate(cash_flows):
        total += cash_flow / (1 + rate) ** year

    return total


# ==========================================
# TEST 1: Future Value
# ==========================================

pv = 1000
rate = 0.05
years = 10

fv = future_value(pv, rate, years)

print("Future Value:", fv)


# ==========================================
# TEST 2: Present Value
# ==========================================

future_cash_flow = 1628.89

pv = present_value(future_cash_flow, rate, years)

print("Present Value:", pv)


# ==========================================
# TEST 3: Net Present Value
# ==========================================

cash_flows = [-1000, 300, 400, 500]

project_npv = npv(0.10, cash_flows)

print("NPV:", project_npv)

if project_npv > 0:
    print("Decision: Positive NPV")
else:
    print("Decision: Negative NPV")
```
