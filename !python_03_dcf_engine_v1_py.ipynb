```python
%%writefile 03_dcf_engine_v1.py

# Simplified DCF Engine v1

# -----------------------------
# INPUTS
# -----------------------------

revenue = 100_000_000_000
operating_margin = 0.30
tax_rate = 0.20
wacc = 0.08
growth_rate = 0.03

debt = 20_000_000_000
cash = 10_000_000_000
shares_outstanding = 15_000_000_000

forecast_years = 5


# -----------------------------
# CALCULATE EBIT
# -----------------------------

ebit = revenue * operating_margin

print("EBIT:", ebit)


# -----------------------------
# CALCULATE AFTER-TAX EBIT
# -----------------------------

after_tax_ebit = ebit * (1 - tax_rate)

print("After-tax EBIT:", after_tax_ebit)


# -----------------------------
# SIMPLIFIED FCF FORECAST
# -----------------------------

fcfs = []

current_fcf = after_tax_ebit

for year in range(1, forecast_years + 1):

    current_fcf = current_fcf * (1 + growth_rate)

    fcfs.append(current_fcf)

    print(f"Year {year} FCF: {current_fcf:,.2f}")


# -----------------------------
# DISCOUNT FUTURE FCF
# -----------------------------

pv_fcfs = []

for year, fcf in enumerate(fcfs, start=1):

    present_value = fcf / ((1 + wacc) ** year)

    pv_fcfs.append(present_value)

    print(f"Year {year} PV of FCF: {present_value:,.2f}")


# -----------------------------
# TERMINAL VALUE
# -----------------------------

terminal_fcf = fcfs[-1] * (1 + growth_rate)

terminal_value = terminal_fcf / (wacc - growth_rate)

print("Terminal Value:", terminal_value)


# -----------------------------
# PRESENT VALUE OF TERMINAL VALUE
# -----------------------------

pv_terminal_value = terminal_value / ((1 + wacc) ** forecast_years)

print("PV of Terminal Value:", pv_terminal_value)


# -----------------------------
# ENTERPRISE VALUE
# -----------------------------

enterprise_value = sum(pv_fcfs) + pv_terminal_value

print("Enterprise Value:", enterprise_value)


# -----------------------------
# EQUITY VALUE
# -----------------------------

equity_value = enterprise_value - debt + cash

print("Equity Value:", equity_value)


# -----------------------------
# INTRINSIC VALUE PER SHARE
# -----------------------------

intrinsic_value_per_share = equity_value / shares_outstanding

print("Intrinsic Value Per Share:", intrinsic_value_per_share)
```
